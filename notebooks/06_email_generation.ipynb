```python
import pandas as pd
```

```python
emails = pd.read_csv("../data/metadata/emails.csv")
df = pd.read_csv("../data/processed/combined_with_clusters.csv")
```

```python
merged_df = pd.merge(emails,df,on="respondent_id",how="inner")
```

```python
def generate_insight(cluster):

    if cluster == "Controlled Spenders":

        return (
            "Your responses suggest relatively disciplined digital spending "
            "behavior with lower financial regret indicators."
        )

    elif cluster == "Routine Evening Spenders":

        return (
            "Your responses indicate routine convenience-driven spending "
            "behavior, especially during evening usage patterns."
        )

    elif cluster == "High Impulsive Spenders":

        return (
            "Your responses showed stronger indicators associated with "
            "impulsive digital spending and post-purchase regret."
        )

    else:

        return (
            "Behavioral spending pattern identified through survey analysis."
        )
```

```python
merged_df["behavioral_insight"] = (
    merged_df["cluster_name"]
    .apply(generate_insight)
)
```

```python
final_df = merged_df[
    [
        "respondent_id",
        "email",
        "regret_intensity",
        "cluster_name",
        "behavioral_insight"
    ]
].copy()
```

```python
final_df = final_df.rename(columns={
    "regret_intensity": "regret_score"
})
```

```python
final_df.to_csv(
    "../data/processed/participant_insights.csv",
    index=False
)
```
