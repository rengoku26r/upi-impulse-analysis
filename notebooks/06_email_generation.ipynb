```python
import pandas as pd
```

```python
emails = pd.read_csv("../data/metadata/emails.csv")
df = pd.read_csv("../data/processed/combined_with_clusters.csv")
```

```python
merged_df = pd.merge(emails,df,on="respondent_id",how="inner")
```

```python
final_df = merged_df[
    [
        "respondent_id",
        "email",
        "regret_intensity",
        "cluster_name",
    ]
].copy()
```

```python
final_df = final_df.rename(columns={
    "regret_intensity": "regret_score"
})
```

```python
final_df.to_csv(
    "../data/processed/participant_insights.csv",
    index=False
)
```
